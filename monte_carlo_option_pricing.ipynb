
# Monte Carlo Simulation – European Call Option Pricing

This notebook demonstrates a basic implementation of Monte Carlo simulation to price a European call option.
The goal is to explain uncertainty modeling over time through a structured simulation process.

```

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```

# Parameters
S0 = 100      # Initial stock price
K = 100       # Strike price
T = 1.0       # Time to maturity (1 year)
r = 0.05      # Risk-free rate
sigma = 0.2   # Volatility
n_simulations = 10000
n_steps = 252
dt = T / n_steps
```

```

# Simulation
S = np.zeros((n_simulations, n_steps))
S[:, 0] = S0

for t in range(1, n_steps):
    z = np.random.standard_normal(n_simulations)
    S[:, t] = S[:, t - 1] * np.exp((r - 0.5 * sigma ** 2) * dt + sigma * np.sqrt(dt) * z)

# Option payoff and price
payoffs = np.maximum(S[:, -1] - K, 0)
call_price = np.exp(-r * T) * np.mean(payoffs)
call_price
```

```

# Plot sample paths
plt.figure(figsize=(10, 5))
for i in range(100):
    plt.plot(S[i], lw=0.5, alpha=0.6)
plt.title("Simulated Stock Price Paths (Monte Carlo)")
plt.xlabel("Time Steps (Days)")
plt.ylabel("Stock Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```
